```python
# Find Duplicate Elements in a List

arr = [1,2,1,3,2,4,5,4]

count = {}
for i in arr:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

dup=[]
for key, value in count.items():
    if value>1:
        dup+=[key]
print(dup)
```

```python
# Return all duplicate elements from a list.
arr = [1,2,1,3,2,4,5,4]

no_duplicate=[]
for i in arr:
    if i not in no_duplicate:
        no_duplicate+=[i]
print(no_duplicate)
```

```python
# Move All Zeros to the End

num = [0,1,0,3,0,12]
j=0

for i in range(len(num)):
    if num[i]!=0:
        num[i],num[j]=num[j],num[i]
        j+=1
print(num)
```

```python
# Move all zeros to the end while maintaining the order of non-zero elements

num = [0,1,0,3,0,12,0]
j=0

for i in range(len(num)):
    if num[i]!=0:
        num[i],num[j]=num[j],num[i]
        j+=1
print(num)
```

```python
# Check if Two Strings are Anagrams

word1='silent'
word2='listen'

def anagram(word1, word2):
    s1=sorted(word1.lower())
    s2=sorted(word2.lower())
    return s1==s2

print(anagram(word1,word2))
```
