```python
# reverse a string without using the inbuilt reverse() method

word = "madamarr"
rev=""

for i in range(len(word)-1,-1,-1):
    rev = rev + word[i]

print(rev)
```

```python
#Anagram or not

word1 = "listen"
word2="silent"

def anagram(word1, word2):
    if sorted(word1) == sorted(word2):
        return True
    else:
        return False

print(anagram(word1, word2))
t1="text"
t2="tent"
print(anagram(t1,t2))
```

```python
# find duplicate elements in an integer array

arr = [1,2,3,4,5,2,4,5]

dict={}

for i in arr:
    if i not in dict:
        dict[i]=1
    else:
        dict[i]+=1
print(dict)

dup = []
for key, value in dict.items():
    if value>1:
        dup+=[key]
print(dup)
```

```python
# find 2nd largest without sorting

num = [10,30,40,70,12,34]

first = num[0]

for i in num:
    if first<i:
        second=first
        first=i

print(second)
```

```python
# move all zeros to end without changing the order

num = [0,1,0,3,0,12]
j=0
for i in range(0,len(num)):
    if num[i]!=0:
        num[i], num[j] = num[j], num[i]
        j+=1

print(num)
```

```python
import asyncio

async def task(name):
    print(f"{name} Start")
    await asyncio.sleep(2)
    print(f"{name} completed")

async def main():
    await asyncio.gather(
        task("hello"),
        task("world")
    )
await main()
```
