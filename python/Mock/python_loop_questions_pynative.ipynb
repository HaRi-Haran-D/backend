```python
# 11. Reverse a string using a for loop (no slicing)

original_str = "Python"
reversed_str = ""

for char in original_str:
    # Prepending: Put the new character before the existing ones
    reversed_str = char + reversed_str

print(f"Original: {original_str}")
print(f"Reversed: {reversed_str}")
```

```python
# 16. Check if a number is a palindrome

num =121
def palindrome(num):
    numb = num
    rev = 0
    while numb > 0:
        rev = rev*10 + numb%10
        numb = numb//10

    return num==rev
print(palindrome(num))
```

```python
# 18. Collatz Conjecture: Generate a sequence until it reaches 1
# The Collatz conjecture states that if you start with any positive integer n, and if n is even,
# divide it by 2; if n is odd, multiply it by 3 and add 1. Repeat the process.
# The sequence will always eventually reach 1.

def collatz(num):
    for i in range(num+1,0,-1):
        if num %2 ==0:
            return 1
```

```python
# 19. Armstrong Number Check

def armstrong(num):
    power = len(str(num))
    numb = num
    rev = 0
    while numb > 0:
        rev = rev + (numb%10)**power
        numb = numb // 10

    return num == rev
print(armstrong(153))
```

```python
arr = [1,2,3,4]
seq = len(arr)-1
new = []

def cumulative_sum(seq):
    if seq == 0:
        return arr[0]
    else:
        return arr[seq] + cumulative_sum(seq-1)

while seq>=0:
    new.append(cumulative_sum(seq))
    seq -= 1

print(new[::-1])
```

```python
# 35. Perfect number check
# Write a program to check if a number is a “Perfect Number.” A perfect number is a positive integer that is equal to the sum of its proper divisors (excluding the number itself). For example, 6 is perfect because 1 + 2 + 3 = 6.
num = 28

def perfect_number(num):
    lst=0
    for i in range(1,(num//2)+1):
        if num % i == 0:
            lst+=i
    if num == lst:
        return "Perfect Number"
    else:
        return "Not Perfect Number"
print(perfect_number(num))
```

```python
# 4: Anagram Checker

word1 = "listen"
word2 = "silent"

def anagram(word1,word2):
    word1 = sorted(word1.lower().replace(" ",""))
    word2 = sorted(word2.lower().replace(" ",""))
    return word1 == word2
print(anagram(word1, word2))
```

```python
# 5: Flatten a Nested List
nested = [1, [2, 3], [4, [5, 6]], 7]
def flatten(arr):
    flat_arr=[]
    for i in arr:
        if isinstance(i,list):
            flat_arr.extend(flatten(i))
        else:
            flat_arr.append(i)
    return flat_arr

print(flatten(nested))
```

```python
# 6: Reverse Each Word of a String

text = "Python is awesome"
def reverse_words(text):
    words = text.split()
    reverse = [word[::-1] for word in words]
    words = " ".join(reverse)
    return words

print(reverse_words(text))
```

```python
# 7. 7: Palindrome Sentence

text = "A man, a plan, a canal: Panama"

def palindrome(text):
    words = ""
    for word in text:
        if word.isalpha():
            words+=word.lower()
    rev = words[::-1]
    if rev == words:
        return "Palindrome"
    else:
        return "Not a Plaindrome"
print(palindrome(text))
```
