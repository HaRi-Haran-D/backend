```python
# 1. Create a function in Python
def show(name,age):
    print(f"Hello {name},you are {age} years old")
show('Hari',25)
```

```python
# 2. Create a function with variable length of arguments
def show(*args):
    for arg in args:
        print(arg)
show(20, 40, 60)
show(80, 100)
```

```python
# 3. Return multiple values from a function
def calculation(a,b):
    add = a+b
    sub = a-b
    return add,sub
print(calculation(40,20))
```

```python
# 4. Create a function with a default argument
def show_employee(name,salary=9000):
    print(f"Hello {name},you are {salary} salary")
show_employee('Hari',15000)
show_employee('Henry')
```

```python
# 5. Create an inner function

def outer(a,b):
    def sum(a,b):
        return a+b
    add = sum(a,b)
    return add+5
print(outer(2,4))
```

```python
# 6. Write a program to create a recursive function that calculates the sum of numbers from 0 to 10.
# Create a recursive function
def sum(num):
    if num <=1:
        return 1
    return num+sum(num-1)
print(sum(10))
```
