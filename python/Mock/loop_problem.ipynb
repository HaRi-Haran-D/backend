```python
# Kadane Problem
arr = [-2,1,-3,4,-1,2,1,-5,4]
max_sum = arr[0]
current_sum = 0
for num in arr:
    current_sum+=num
    if current_sum > max_sum:
        max_sum = current_sum
    if current_sum < 0:
        current_sum = 0
print(max_sum)
```

```python
# 2. Write a python program to calculate the number of days between two dates.
d1 = "2014-7-2"
d2 = "2014-7-11"

from datetime import datetime
def calculate(d1,d2):
    d1= datetime.strptime(d1, '%Y-%m-%d')
    d2= datetime.strptime(d2, '%Y-%m-%d')
    diff = (d2 - d1).days
    return diff
print(calculate(d1,d2))
```

```python
# 5. write a program that reverse and replace odd digits with 0
num = 12345
# output = 40200

def reverse_and_replace(num):
    text = str(num)
    rev =text[::-1]

    if len(rev)<2:
        return rev
    else:
        for i in range(0,len(rev)):
            if int(rev[i])% 2!=0:
                rev[i]=0
            else:
                rev[i],rev[i-1] = rev[i-1],rev[i]
    return rev
print(reverse_and_replace(num))
```

```python
def reverse_and_replace(num):
    text = str(num)[::-1]
    print("Reversed:", text)   # Debug
    result = ""
    for ch in text:
        if int(ch) % 2 != 0:
            result += "0"
        else:
            result += ch
        print("Step:", result)  # Debug
    return result

print(reverse_and_replace(12345))
```

```python
num = 12345

def reverse_integer(num):
    reverse = 0
    while num>0:
        reminder = num % 10
        reverse = (reverse * 10) + reminder
        num = num // 10
    return reverse
print(reverse_integer(num))
```

```python
d1 = {'a':10,'b':20}
d2 = {'b':15,'c':25}
def merge(d1,d2):
    result = d1.copy()
    for key,  value in d2.items():
        result[key] = result.get(key,0)+value
    return result
print(merge(d1,d2))
```

```python
text = "Python Programming"
from collections import Counter
count = Counter(text)
print(count)
text = text.replace(" ","")
def counting(text):
    count = {}
    for i in text:
        if i in count:
            count[i]+=1
        else:
            count[i]=1
    return count

print(counting(text))
```
