```python
# Kadane Problem
arr = [-2,1,-3,4,-1,2,1,-5,4]
max_sum = arr[0]
current_sum = 0
for num in arr:
    current_sum+=num
    if current_sum > max_sum:
        max_sum = current_sum
    if current_sum < 0:
        current_sum = 0
print(max_sum)
```

```python
num = 4567
rev = 0
while num>0:
    rem = num % 10
    rev = (rev * 10) + rem
    num = num // 10
print(rev)
```

```python
a= 10
def prime(num):
    if num < 2:
        return False
    for i in range(2,int((num**0.5)+1)):
        if num % i == 0:
            return False
    return True

for k in range(2,50):
    if prime(k):
        print(k)
```
