```python
num = 9875
# output:2

sum=0

while num > 10:
    for i in str(num):
        sum += int(i)
    num = sum
    sum=0

print(num)
```

```python
num = [1,2,2,3,3,3,4,4]
#second most frequent

dict ={}
for i in num:
    if i not in dict:
        dict[i]=1
    else:
        dict[i]+=1


values = sorted(set(dict.values()), reverse=True)
second_highest = values[1]

for key, value in dict.items():
    if value == second_highest:
        print(key)
```
