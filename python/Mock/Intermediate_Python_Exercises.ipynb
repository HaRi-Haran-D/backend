```python
# 1. List Comprehension Mastery
# Write a single-line list comprehension that takes a list of strings, filters out strings shorter than
# 4 characters, and converts the remaining strings to uppercase.

words = ["apple", "bat", "cherry", "dog", "elderberry"]
# output = ['APPLE', 'CHERRY', 'ELDERBERRY']
new=[]
for word in words:
    if len(word) < 4:
        continue
    else:
        new.append(word.upper())
print(new)

new = [word.upper() for word in words if len(word)> 4]
print(new)
```

```python
# 2. Dictionary Merging with Logic
# Write a function that merges two dictionaries. If a key exists in both dictionaries, sum their values. If a key exists in only one, include it as is.

dict_a = {'a': 10, 'b': 20}
dict_b = {'b': 5, 'c': 15}
# Output: Merged Dictionary: {'a': 10, 'b': 25, 'c': 15}
def merge(a,b):
    result = a.copy()
    for i,j in b.items():
        result[i] = result.get(i,0)+j
    return result
print(merge(dict_a,dict_b))
```

```python
# 3. Frequency Map with Counter
# Create a function that takes a string and returns a count of how many times each character appears. Ignore spaces and make it case-insensitive.

text = "Python Programming"
from collections import Counter

def count(text):
    txt = text.lower().replace(" ","")
    return Counter(txt)
print(count(text))
```

```python
# 4. Anagram Checker
word1 = "listen"
word2 = "silent"

def anagram(word1, word2):
    if len(word1) != len(word2):
        return False
    elif sorted(word1.lower().replace(" ","")) == sorted(word2.lower().replace(" ","")):
        return True
print(anagram(word1, word2))
```

```python
# 5. Flatten a Nested List
# Write a recursive function that takes a list containing other lists (of any depth) and returns a single “flat” list of all elements.

arr = [[10,11], [2, 3], [4, [5, 6]], 7]
# output [1, 2, 3, 4, 5, 6, 7]

def flatten(arr):
    new = []
    for item in arr:
        if isinstance(item,list):
            new.extend(flatten(item))
        else:
            new+=[item]
    return new
arr = [1, [2, 3], [4, [5, 6]], [7,8]]
print(flatten(arr))


# def flatten(arr):
#     new = []
#     for item in arr:
#         if isinstance(item,list):
#             new.extend(flatten(item))
#         else:
#             new.append(item)
#     return new
# arr = [1, [2, 3], [4, [5, 6]], [7,8]]
# print(flatten(arr))
```

```python
# 6. Reverse Each Word of a String
#  Given a sentence, reverse each individual word within the string while maintaining the original word order.

text = "Python is awesome"
# output "nohtyP si emosewa"
def revere_str(text):
    stri = text.split()
    reverse = [word[::-1] for word in stri]
    out = " ".join(reverse)
    print(out)
revere_str(text)
```

```python
# 7. Palindrome Sentence
# Write a function to check if a full sentence is a palindrome. You must ignore case, spaces, and all punctuation marks.
text = "A man, a plan, a canal: Panama"

def palin(text):
    sent=""
    for word in text:
        if word.isalpha():
            sent+=word.lower()
    rev = sent[::-1]
    if rev == sent:
        return True
    else:
        return False
print(palin(text))
```

```python
# 8. List Comprehension Filtering (Advanced)
# Given a list of strings, use a single list comprehension to extract strings that meet two criteria: they must be longer than 5 characters AND they must start with a vowel (a, e, i, o, u).

arr = ["apple", "education", "ice", "ocean", "python", "umbrella"]
def strings(arr):
    vowels = ["a","e","i","o","u"]

    new = [word for word in arr if len(word) > 5 and word[0].lower() in vowels]
    return new
print(strings(arr))
```

```python
# 9. Remove Duplicates (Preserving Order)
# Write a function that removes duplicate elements from a list. You cannot use set() because sets do not maintain the original order of elements.
arr = [1, 2, 2, 3, 1, 4, 2]
def remove_duplicates(arr):
    new = []
    for num in arr:
        if num not in new:
            new.append(num)
    return new
print(remove_duplicates(arr))
```

```python
# 10. Circular Shift (Rotation)
# Create a function rotate_list(lst, n, direction) that shifts the elements of a list by N positions. The direction can be ‘left’ or ‘right’.

arr = [1, 2, 3, 4, 5]
# output [4, 5, 1, 2, 3] Shift: 2, Direction: 'right'
def rotate_list(lst, n, direction='right'):
    if not lst:
        return lst

    # Handle shifts larger than the list length
    n = n % len(lst)

    if direction == 'right':
        # Take the last n elements and put them at the front
        return lst[-n:] + lst[:-n]
    else:
        # Take the first n elements and put them at the back
        return lst[n:] + lst[:n]

data = [1, 2, 3, 4, 5]
print(f"Right Shift 2: {rotate_list(data, 2, 'right')}")
print(f"Left Shift 1:  {rotate_list(data, 1, 'left')}")
```

```python
# 11. Dictionary Merging (Value Grouping)
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
# output {'a': [1], 'b': [2, 3], 'c': [4]}
def merge_and_group(dict1, dict2):
    combined = {}
    all_keys = set(dict1.keys()) | set(dict2.keys())

    for key in all_keys:
        values = []
        if key in dict1:
            values.append(dict1[key])
        if key in dict2:
            values.append(dict2[key])
        combined[key] = values

    return combined
print(merge_and_group(d1, d2))
```

```python
# 12. Inverted Index
d1 = {"Orwell": ["1984", "Animal Farm"], "Huxley": ["Brave New World"]}
# output {'1984': 'Orwell', 'Animal Farm': 'Orwell', 'Brave New World': 'Huxley'}

def invert(data):
    inverted={}
    for author, books in data.items():
        for book in books:
            inverted[book] = author
    return inverted

print(invert(d1))
```

```python
# 13. Dictionary Sorting (Lambda)
employees = [{"name": "A", "salary": 50}, {"name": "B", "salary": 70}, {"name": "C", "salary": 60}]
# output [{'name': 'B', 'salary': 70}, {'name': 'C', 'salary': 60}, {'name': 'A', 'salary': 50}]

n = sorted(employees, key = lambda x: x['salary'], reverse=True)
print(n)
```

```python
# 14. Subset and Superset Validation
set1 = {1, 2, 3}
set2 = {1,2,3,4,5}

def relation(set1,set2):
    if set1.issubset(set2):
        print("Subset")
    elif set1.issuperset(set2):
        print("Superset")
    if set1.isdisjoint(set2):
        print("Disjoint")
    else:
        print("share Elements")
relation(set1,set2)
```
