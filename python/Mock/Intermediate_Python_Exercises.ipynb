```python
# 1. List Comprehension Mastery
# Write a single-line list comprehension that takes a list of strings, filters out strings shorter than
# 4 characters, and converts the remaining strings to uppercase.

words = ["apple", "bat", "cherry", "dog", "elderberry"]
# output = ['APPLE', 'CHERRY', 'ELDERBERRY']
new=[]
for word in words:
    if len(word) < 4:
        continue
    else:
        new.append(word.upper())
print(new)

new = [word.upper() for word in words if len(word)> 4]
print(new)
```

```python
# 2. Dictionary Merging with Logic
# Write a function that merges two dictionaries. If a key exists in both dictionaries, sum their values. If a key exists in only one, include it as is.

dict_a = {'a': 10, 'b': 20}
dict_b = {'b': 5, 'c': 15}
# Output: Merged Dictionary: {'a': 10, 'b': 25, 'c': 15}
def merge(a,b):
    result = a.copy()
    for i,j in b.items():
        result[i] = result.get(i,0)+j
    return result
print(merge(dict_a,dict_b))
```

```python
# 3. Frequency Map with Counter
# Create a function that takes a string and returns a count of how many times each character appears. Ignore spaces and make it case-insensitive.

text = "Python Programming"
from collections import Counter

def count(text):
    txt = text.lower().replace(" ","")
    return Counter(txt)
print(count(text))
```

```python
# 4. Anagram Checker
word1 = "listen"
word2 = "silent"

def anagram(word1, word2):
    if len(word1) != len(word2):
        return False
    elif sorted(word1.lower().replace(" ","")) == sorted(word2.lower().replace(" ","")):
            return True
print(anagram(word1, word2))
```

```python
# 5. Flatten a Nested List
# Write a recursive function that takes a list containing other lists (of any depth) and returns a single “flat” list of all elements.

arr = [1, [2, 3], [4, [5, 6]], 7]
# output [1, 2, 3, 4, 5, 6, 7]

def flatten(arr):
    new = []
    for item in arr:
        if isinstance(item,list):
            new.extend(flatten(item))
        else:
            new+=[item]
    return new
arr = [1, [2, 3], [4, [5, 6]], [7,8]]
print(flatten(arr))
```

```python
# 6. Reverse Each Word of a String
#  Given a sentence, reverse each individual word within the string while maintaining the original word order.

text = "Python is awesome"
# output "nohtyP si emosewa"
def revere_str(text):
    stri = text.split()
    reverse = [word[::-1] for word in stri]
    out = " ".join(reverse)
    print(out)
revere_str(text)
```

```python
# 7. Palindrome Sentence
# Write a function to check if a full sentence is a palindrome. You must ignore case, spaces, and all punctuation marks.
text = "A man, a plan, a canal: Panama"

def palin(text):
    sent=""
    for word in text:
        if word.isalpha():
            sent+=word.lower()
    rev = sent[::-1]
    if rev == sent:
        return True
    else:
        return False
print(palin(text))
```

```python
# 8. List Comprehension Filtering (Advanced)
# Given a list of strings, use a single list comprehension to extract strings that meet two criteria: they must be longer than 5 characters AND they must start with a vowel (a, e, i, o, u).

arr = ["apple", "education", "ice", "ocean", "python", "umbrella"]
def strings(arr):
    vowels = ["a","e","i","o","u"]

    new = [word for word in arr if len(word) > 5 and word[0].lower() in vowels]
    return new
print(strings(arr))
```

```python
# 9. Remove Duplicates (Preserving Order)
# Write a function that removes duplicate elements from a list. You cannot use set() because sets do not maintain the original order of elements.
arr = [1, 2, 2, 3, 1, 4, 2]
def remove_duplicates(arr):
    new = []
    for num in arr:
        if num not in new:
            new.append(num)
    return new
print(remove_duplicates(arr))
```
